### 🤖 Fase 5: Integração de Modelos Reais (.pkl)
Nesta etapa, vamos simular a exportação de um modelo treinado e atualizar a API para consumi-lo.

````python
import joblib
import numpy as np
from sklearn.ensemble import RandomForestRegressor
import os

# Criando pasta para os modelos
os.makedirs('backend/models', exist_ok=True)

# Criando e treinando um modelo dummy (Dívida/PIB ~ Risco Brasil + Juro Real)
X_train = np.random.rand(100, 2)  # Features: [Risco, Juros]
y_train = 60 + 0.05 * X_train[:, 0] + 2 * X_train[:, 1]

model = RandomForestRegressor(n_estimators=10)
model.fit(X_train, y_train)

# Salvando o modelo
model_path = 'backend/models/model_v7.pkl'
joblib.dump(model, model_path)

print(f"Modelo de teste salvo em: {model_path}")
````

````python
flask_app_updated = """from flask import Flask, jsonify, request
import joblib
import numpy as np
import os

app = Flask(__name__)

# Carregamento do modelo na inicialização
MODEL_PATH = 'backend/models/model_v7.pkl'
if os.path.exists(MODEL_PATH):
    model = joblib.load(MODEL_PATH)
else:
    model = None

@app.route('/')
def index():
    return jsonify({
        'project': 'MacroLab v7',
        'model_loaded': model is not None,
        'status': 'online'
    })

@app.route('/predict', methods=['POST', 'GET'])
def predict():
    if not model:
        return jsonify({'error': 'Modelo não encontrado'}), 500

    # Exemplo de entrada: ?risco=200&juros=5
    try:
        risco = float(request.args.get('risco', 250))
        juros = float(request.args.get('juros', 6))

        # Preparar dados para o modelo (2D array)
        input_data = np.array([[risco, juros]])
        prediction = model.predict(input_data)[0]

        return jsonify({
            'input': {'risco': risco, 'juros': juros},
            'prediction_divida_pib': round(float(prediction), 2)
        })
    except Exception as e:
        return jsonify({'error': str(e)}), 400

if __name__ == '__main__':
    app.run(debug=True, port=8081)
"""

with open('backend/app.py', 'w') as f:
    f.write(flask_app_updated)

print("Backend atualizado com suporte a carregamento de modelo .pkl")
````

### 🧪 Teste de Cenários Macroeconômicos
Este script testa a API com diferentes combinações de entrada para simular condições de mercado variadas.

````python
import requests
import pandas as pd
import threading
import time
from backend.app import app

def run_server():
    # Rodando na porta 8082 conforme configurado nos testes de modelo
    app.run(port=8082, debug=False, use_reloader=False)

def test_scenarios():
    # Iniciando o servidor localmente para o teste
    server_thread = threading.Thread(target=run_server)
    server_thread.daemon = True
    server_thread.start()
    time.sleep(2) # Aguarda o boot do servidor

    base_url = 'http://127.0.0.1:8082'

    # Definindo cenários [Risco Brasil, Juro Real]
    scenarios = {
        'Otimista': {'risco': 150, 'juros': 3.5},
        'Neutro': {'risco': 250, 'juros': 6.0},
        'Pessimista': {'risco': 500, 'juros': 12.0}
    }

    results = []
    print("--- Testando Cenários no MacroLab v7 ---")

    for name, params in scenarios.items():
        try:
            response = requests.get(f'{base_url}/predict', params=params)
            if response.status_code == 200:
                pred = response.json()['prediction_divida_pib']
                results.append({
                    'Cenário': name,
                    'Risco (bps)': params['risco'],
                    'Juros (%)': params['juros'],
                    'Prev. Dívida/PIB (%)': pred
                })
                print(f"✅ {name}: {pred}%")
            else:
                print(f"❌ Erro no cenário {name}: {response.status_code}")
        except Exception as e:
            print(f"⚠️ Falha de conexão: {e}")

    if results:
        df_results = pd.DataFrame(results)
        print("\nResumo dos Cenários:")
        display(df_results)

if __name__ == '__main__':
    test_scenarios()
````
